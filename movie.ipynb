```python
import numpy as np 
import pandas as pd
```

```python
movie=pd.read_csv('tmdb_5000_movies.csv')
credit=pd.read_csv('tmdb_5000_credits.csv')
```

```python
movie.sample(1)
```

```python
credit.sample(2)
```

```python
df=movie.merge(credit,on='title')
```

```python
df.sample(1)
```

```python
# columns that we keep in our dataset
# generes 
# id
# keyword 
# overview
# cast
#crew
df=df[['id','genres','keywords','overview','cast','crew']]
```

```python
df.sample(1)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
```

```python
df.duplicated().sum()
```

```python
import ast 
def converter(obj):
    l=[]
    for i in ast.literal_eval(obj):
        l.append(i['name'])
    return l
```

```python
df['genres']=df['genres'].apply(converter)
```

```python
df.head(1)
```

```python
df['keywords'][0]
```

```python
df['keywords']=df['keywords'].apply(converter)
```

```python
df.head(1)
```

```python
df['cast'][0]
```

```python
def converter2(obj):
    l=[]
    count=0
    for i in  ast.literal_eval(obj):
        if count<=3:
           l.append(i['name'])
           count+=1
        else:
            break
    return l
```

```python
df['cast']=df['cast'].apply(converter2)
```

```python
df.head(1)
```

```python
df['cast'][0]
```

```python
df['crew'][2]
```

```python
def converter3(obj):
    l=[]
    for i in ast.literal_eval(obj):
        if i['job']=="Director":
            l.append(i['name'])
            break
    return l 
```

```python
df['crew']=df['crew'].apply(converter3)
```

```python
df.head(1)
```

```python
df['overview']=df['overview'].apply(lambda x:x.split())
```

```python
df.head(5)
```

```python
# romoving extra spaces
df['genres']=df['genres'].apply(lambda x:[i.replace(" ","") for i in x])
df['keywords']=df['keywords'].apply(lambda x:[i.replace(" ","") for i in x])
df['cast']=df['cast'].apply(lambda x:[i.replace(" ","") for i in x])
df['crew']=df['crew'].apply(lambda x:[i.replace(" ","") for i in x])
```

```python
df.head()
```

```python
df['tag']=df['genres'] + df['keywords'] + df['overview'] + df['cast'] + df['crew']
```

```python
df.head(1)
```

```python
new_df=df[['id','tag']]
```

```python
new_df.head(2)
```

```python
new_df['title']=movie['title']
```

```python
new_df
```

```python
new_df.isnull().sum()
```

```python
new_df.dropna(inplace=True)
```

```python
new_df['tag']=new_df['tag'].apply(lambda x:" ".join(x))
```

```python
new_df
```

```python
new_df['tag']=new_df['tag'].apply(lambda x:x.lower())
```

```python
new_df
```

```python
# steamming 
import nltk
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def stem(text):
    l=[]
    for i in text.split():
        l.append(ps.stem(i))
    return " ".join(l)
```

```python
new_df['tag']=new_df['tag'].apply(stem)
```

```python
new_df['tag'][0]
```

```python
# vectorization 
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
```

```python
vector=cv.fit_transform(new_df['tag']).toarray()
```

```python
vector.shape
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_score=cosine_similarity(vector)
```

```python
similarity_score[2]
```

```python
def recommend(movie):
    index=new_df[new_df['title']==movie].index[0]
    distance=similarity_score[index]
    recommend_movie=sorted(list(enumerate(distance)),reverse=True,key=lambda x:x[1])[1:6]
    for i in recommend_movie:
        print(new_df.iloc[i[0]].title)
    
    
    
```

```python
recommend("Pirates of the Caribbean: At World's End")
```
